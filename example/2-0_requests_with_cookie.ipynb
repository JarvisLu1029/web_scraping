### Requests 夾帶 cookie

```python
import requests
from bs4 import BeautifulSoup as bs

 # PTT Gossiiping (八卦版) 
url = "https://www.ptt.cc/bbs/Gossiping/index.html"

# 設定 cookie
my_cookies = {
    "over18": "1"
}

# 用 requests 的 get 方法把網頁抓下來
res = requests.get(url, cookies=my_cookies)

# 1. 送出去的 cookie
print("=== Request Cookies ===")
print(res.request.headers.get("Cookie"))

# 2. server 回來的 cookie
print("\n=== Response Cookies ===")
print(res.cookies)

# 指定 lxml 作為解析器
soup = bs(res.text, "lxml")
# print(res.text)

# 首頁網址
prefix = 'https://www.ptt.cc'

# 顯示連結列表
for a in soup.select('div.r-ent > div.title > a'):
    print(a.get_text())
    print(prefix + a['href'])
```

```python
from selenium import webdriver

driver = webdriver.Chrome()

url = "https://www.ptt.cc/bbs/Gossiping/index.html"

driver.get(url)

# 設定 cookie
driver.add_cookie({
    "name": "over18",
    "value": "1"
})

# 設定完 cookie 後，重新載入頁面
driver.get(url)
```
